```python
#------------------------------------------------------------------------------------------------
# PROBLEM DEFINITION IN PYOMO (3/4)
#------------------------------------------------------------------------------------------------

#------------------------------------------------------------------------------------------------
# CONSTRAINTS
#------------------------------------------------------------------------------------------------

# BESS MODEL CONSTRAINTS ------------------------------------------------------------------------
# BESS discharging power (pb_dch[t] <= Pb_dch_max)
def bess_dch_limits(m, b, s, t):
    return m.pb_dch[b, s, t] <= m.Pb_dch_max[b] * m.xb[b, s, t]
m.PbessDischarge = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_dch_limits)

# BESS charging power (pb_ch[t] <= Pb_ch_max)
def bess_ch_limits(m, b, s, t):
    return m.pb_ch[b, s, t] <= m.Pb_ch_max[b] * (1 - m.xb[b, s, t])
m.PbessCharge = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_ch_limits)

# BESS SoC modelling (SoC[0] = ...; SoC[t] = (1 - sig_b_sd) * SoC[t-1] + (1/Eb_rtd) * (eta_bch * pb_ch - pb_dch/eta_bdch) * Dt)
def soc_eval(m, b, s, t):
    if t == m.TIME.first():
        return m.soc[b, s, 0] == (1 - m.sig_b_sd[b]) * m.SoC_0[b] + (1 / m.Eb_rtd[b]) * (m.eta_bch[b] * m.pb_ch[b, s, 0] - m.pb_dch[b, s, 0] / m.eta_bdch[b]) * m.Dt
    else:
        return m.soc[b, s, t] == (1 - m.sig_b_sd[b]) * m.soc[b, s, t-1] + (1 / m.Eb_rtd[b]) * (m.eta_bch[b] * m.pb_ch[b, s, t] - m.pb_dch[b, s, t] / m.eta_bdch[b]) * m.Dt
m.SoCEvaluation = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = soc_eval)

# End SoC condition on the BESS (SoC[end] == SoC_0)
def soc_end(m, b, s):
    return m.soc[b, s, m.TIME.last()] == m.SoC_0[b]
m.SoCEnd = pyo.Constraint(m.BESS, m.SCEN, rule = soc_end)


# EL MODEL CONSTRAINTS --------------------------------------------------------------------------
# EL working states selection (xel_on[t] + xel_off[t] + xel_sb[t] = 1)
def el_states(m, e, s, t):
    return m.xel_on[e, s, t] + m.xel_off[e, s, t] + m.xel_sb[e, s, t] == 1
m.ELStates = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_states)

# Identification of EL startup (xel_su[0] = 0; xel_su[t] >= xel_off[t-1] + xel_on[t] + xel_sb[t] - 1)
def el_find_startup(m, e, s, t):
    if t == m.TIME.first():
        return m.xel_su[e, s, 0] == 0 
    else:
        return m.xel_su[e, s, t] >= m.xel_off[e, s, t-1] + m.xel_on[e, s, t] + m.xel_sb[e, s, t] - 1
m.ELFindStartup = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_find_startup)

# Hydrogen-power EL piecewise linear characteristic    # REPLACED BY EXPRESSION

# EL power boundaries during on state
def el_on_bound_low(m, e, s, t):
    return m.Pel_min[e] * m.xel_on[e, s, t] <= m.pel_on[e, s, t]
m.ELOnBoundLow = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule=el_on_bound_low)

def el_on_bound_high(m, e, s, t):
    return m.pel_on[e, s, t] <= m.Pel_rtd[e] * m.xel_on[e, s, t]
m.ELOnBoundHigh = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule=el_on_bound_high)

# EL power evaluation    # REPLACED BY EXPRESSION

# EL ramp rate constraints
def el_rampup(m, e, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip 
    return m.rho_elneg[e] <= m.pel[e, s, t+1] - m.pel[e, s, t]
m.ELrampup = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_rampup)
def el_rampdown(m, e, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip
    return m.pel[e, s, t+1] - m.pel[e, s, t] <= m.rho_elpos[e]
m.ELrampdown = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_rampdown)


# HSS MODEL CONSTRAINTS -------------------------------------------------------------------------
# Hydrogen tank level modelling
def h2_tank_level(m, hs, s, t):
    if t == m.TIME.first():
        return m.hhss[hs, s, t] == (1 - m.sig_hss_sd[hs]) * m.H_hss_0[hs] + \
            (1 - m.sig_hss_ch[hs]) * (1 - m.sig_hc_el2hss[hs]) * sum(m.hel[e, s, t] for e in m.EL) \
            - sum(m.hfc[f, s, t] for f in m.FC) / (1 - m.sig_hss_dch[hs]) - m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs]))
    else:
        return m.hhss[hs, s, t] == (1 - m.sig_hss_sd[hs]) * m.hhss[hs, s, t-1] + \
            (1 - m.sig_hss_ch[hs]) * (1 - m.sig_hc_el2hss[hs]) * sum(m.hel[e, s, t] for e in m.EL) \
            - sum(m.hfc[f, s, t] for f in m.FC) / (1 - m.sig_hss_dch[hs]) - m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs]))
m.H2TankLevel = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = h2_tank_level)

# End hydrogen tank level condition
def h2_tank_level_end(m, hs, s):
    return m.hhss[hs, s, m.TIME.last()] == m.H_hss_0[hs]
m.H2TankLevelEnd = pyo.Constraint(m.HSS, m.SCEN, rule = h2_tank_level_end)

# Maximum compressor flow rate el2hss
def max_comprflow_el2hss(m, hs, s, t):
    return sum(m.hel[e, s, t] for e in m.EL) <= m.FR_hc_el2hss[hs]
m.MaxComprFlowEl2hss = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = max_comprflow_el2hss)

# Maximum compressor flow rate hss2dem
def max_comprflow_hss2dem(m, hs, s, t):
    return m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs])) <= m.FR_hc_hss2dem[hs]
m.MaxComprFlowHss2dem = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = max_comprflow_hss2dem)


# FC MODEL CONSTRAINTS --------------------------------------------------------------------------
# Hydrogen-power linear model   # REPLACED BY EXPRESSION

# FC on state power boundaries
def fc_on_bound_low(m, f, s, t):
    return m.Pfc_min[f] * m.xfc_on[f, s, t] <= m.pfc[f, s, t]
m.FCOnBoundLow = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_on_bound_low)
def fc_on_bound_high(m, f, s, t):
    return m.pfc[f, s, t] <= m.Pfc_rtd[f] * m.xfc_on[f, s, t]
m.FCOnBoundHigh = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_on_bound_high)

# FC ramp rate constraints
def fc_rampup(m, f, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip 
    return m.rho_fcneg[f] <= m.pfc[f, s, t+1] - m.pfc[f, s, t]
m.FCrampup = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_rampup)
def fc_rampdown(m, f, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip
    return m.pfc[f, s, t+1] - m.pfc[f, s, t] <= m.rho_fcpos[f]
m.FCrampdown = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_rampdown)


# FCR MODELLING ---------------------------------------------------------------------------------
# EL FCR Modelling -------------------------------------------------------
# EL FCR participation limits based on power capacity
def el_fcr_caplim_down(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.Pel_rtd[e] * m.xel_on[e, s, t] - m.pel_on[e, s, t]
m.ELFCRCapacityLimDown = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_down)

def el_fcr_caplim_up(m, e, s, t):
    return m.pel_fcr_up[e, s, t] <= m.pel_on[e, s, t] - m.Pel_min[e] * m.xel_on[e, s, t]
m.ELFCRCapacityLimUp = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_up)

# EL FCR participation limits based on ramp rate
def el_fcr_rampratelim_down(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.rho_elpos[e] * m.DT_FCR_ramp
m.ELFCRRamprateLimDown = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_rampratelim_down)

def el_fcr_rampratelim_up(m, e, s, t):
    return m.pel_fcr_up[e, s, t] <= -m.rho_elneg[e] * m.DT_FCR_ramp
m.ELFCRRamprateLimUp = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_rampratelim_up)

# EL FCR participation limit based on available renewable energy
# If I wanted to generalize this constraint to multiple ELs, the left-hand side would have to be the sum over all the ELs of the available
# capacity for FCR. In other words, the constraint would determine the total available energy for FCR by the ELs, which would then have to
# be divided amongst all the ELs (meaning: sum(m.pel_fcr_dwn[e, s, t] for i in m.EL) <= left-hand side of the constraint below).
def el_fcr_availab_pow(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.Ppv[s, t] + sum(m.pb_dch[b, s, t] - m.pb_ch[b, s, t] for b in m.BESS) - sum(m.pel_on[e, s, t] for e in m.EL) \
                                   - sum(m.phc_el2hss[e, s, t] for e in m.EL) - sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS)
m.ELFCRAvailabPower = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_availab_pow)

# EL FCR participation limit based on available hydrogen storage capacity
def el_fcr_availab_h2(m, e, s, t):
    return m.A[e] * m.pel_fcr_dwn[e, s, t] <= sum(m.H_hss_max[hs] - m.hhss[hs, s, t] for hs in m.HSS)
m.ELFCRAvailabH2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_availab_h2)

if sym_fcr:
    # EL FCR symmetric capacity evaluation
    def el_fcr_sym(m, e, s, t):
        return m.pel_fcr_dwn[e, s, t] == m.pel_fcr_up[e, s, t]
    m.ELFCRSymmetric = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_sym)

# FC FCR Modelling -------------------------------------------------------
# FC FCR participation limits based on power capacity
def fc_fcr_caplim_down(m, f, s, t):
    return m.pfc_fcr_dwn[f, s, t] <= m.pfc[f, s, t] - m.Pfc_min[f] * m.xfc_on[f, s, t]
m.FCFCRCapacityLimDown = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_down)

def fc_fcr_caplim_up(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] <= m.Pfc_rtd[f] * m.xfc_on[f, s, t] - m.pfc[f, s, t]
m.FCFCRCapacityLimUp = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_up)

# FC FCR participation limits based on ramp rate
def fc_fcr_rampratelim_down(m, f, s, t):
    return m.pfc_fcr_dwn[f, s, t] <= -m.rho_fcneg[f] * m.DT_FCR_ramp
m.FCFCRRamprateLimDown = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_rampratelim_down)

def fc_fcr_rampratelim_up(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] <= m.rho_fcpos[f] * m.DT_FCR_ramp
m.FCFCRRamprateLimUp = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_rampratelim_up)

# FC FCR participation limit based on hydrogen availability
def fc_fcr_availab_h2(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] / ((1-m.sig_hss_dch['HSS']) * m.eta_fc[f] * m.lhv) <= sum(m.hhss[hs, s, t] - m.H_hss_min[hs] for hs in m.HSS)
m.FCFCRAvailabH2 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_availab_h2)

if sym_fcr:
    # FC FCR symmetric capacity evaluation
    def fc_fcr_sym(m, f, s, t):
        return m.pfc_fcr_dwn[f, s, t] == m.pfc_fcr_up[f, s, t]
    m.FCFCRSymmetric = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_sym)


# SYSTEM CONSTRAINTS ----------------------------------------------------------------------------
# Grid power limits
def grid_power_importlim(m, s, t):
    return m.pgrid_p[s, t] <= m.Pgrid_max_import * m.xgrid[s, t]
m.GridPowerImportLim = pyo.Constraint(m.SCEN, m.TIME, rule=grid_power_importlim)

def grid_power_exportlim(m, s, t):
    return m.pgrid_n[s, t] <= m.Pgrid_max_export * (1 - m.xgrid[s, t])
m.GridPowerExportLim = pyo.Constraint(m.SCEN, m.TIME, rule=grid_power_exportlim)


# Green hydrogen constraint
def green_h2(m, s, t):
    return sum(m.pel[e, s, t] for e in m.EL) + sum(m.pb_ch[b, s, t] for b in m.BESS) \
    + sum(m.phc_el2hss[e, s, t] for e in m.EL) + sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS) <= m.Ppv[s, t] + sum(m.pb_dch[b, s, t] for b in m.BESS)
m.GreenHydrogen = pyo.Constraint(m.SCEN, m.TIME, rule = green_h2)


# General power balance
def power_balance(m, s, t):
    return m.pgrid_p[s, t] - m.pgrid_n[s, t] + m.Ppv[s, t] - m.Pd_load[s, t] \
    + sum(m.pb_dch[b, s, t] - m.pb_ch[b, s, t] for b in m.BESS) \
    - sum(m.pel[e, s, t] for e in m.EL) + sum(m.pfc[f, s, t] for f in m.FC) \
    - sum(m.phc_el2hss[e, s, t] for e in m.EL) - sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS) == 0
m.PowerBalance = pyo.Constraint(m.SCEN, m.TIME, rule = power_balance)


# HYDROGEN DEMAND CONSTRAINTS -------------------------------------------------------------------
# Daily limits on hydrogen supply
def h2_daily_supply_bound_low(m, s, d):
    return m.Hsup_min_day <= sum(m.hdem[s, d*24 + h] for h in range(24))
m.H2DailySupplyBoundLow = pyo.Constraint(m.SCEN, m.DAYS, rule = h2_daily_supply_bound_low)
def h2_daily_supply_bound_high(m, s, d):
    return sum(m.hdem[s, d*24 + h] for h in range(24)) <= m.Hsup_max_day
m.H2DailySupplyBoundHigh = pyo.Constraint(m.SCEN, m.DAYS, rule = h2_daily_supply_bound_high)

# Weekly / Monthly hydrogen supply target
def h2_target_supply(m, s, w):
    return sum(m.hdem[s, w*24*7 + h] for h in range(24*7)) == m.Hsup_tgt_week
m.H2TargetSupply = pyo.Constraint(m.SCEN, m.WEEKS, rule = h2_target_supply)
```

```python
#------------------------------------------------------------------------------------------------
# PROBLEM DEFINITION IN PYOMO (4/4)
#------------------------------------------------------------------------------------------------

#------------------------------------------------------------------------------------------------
# OBJECTIVE FUNCTION
#------------------------------------------------------------------------------------------------

def tot_revenue(m):
    return (1 / m.num_scen) * sum(m.hdem[s, t] * m.C_h2 \
                + m.pgrid_n[s, t] * m.Dt * m.C_ele_sell[s, t] \
                - m.pgrid_p[s, t] * m.Dt * m.C_ele[s, t] \
                + sum(m.pel_fcr_up[e, s, t] * m.C_FCR_capup + m.pel_fcr_dwn[e, s, t] * m.C_FCR_capdown for e in m.EL) \
                + sum(m.pfc_fcr_up[f, s, t] * m.C_FCR_capup + m.pfc_fcr_dwn[f, s, t] * m.C_FCR_capdown for f in m.FC) for s in m.SCEN for t in m.TIME)
m.Obj = pyo.Objective(rule = tot_revenue, sense = pyo.maximize)

# def tot_revenue(m):
#     return (1 / m.num_scen) * sum(m.hdem[s, t] * m.C_h2 \
#                 + (m.Ppv[s, t] + sum(m.pb_dch[b, s, t] - m.pb_ch[b, s, t] for b in m.BESS) \
#                 - sum(m.pel[e, s, t] for e in m.EL) + sum(m.pfc[f, s, t] for f in m.FC)) * m.Dt * m.C_ele[s, t] \
#                 + sum(m.pel_fcr_up[e, s, t] * m.C_FCR_capup + m.pel_fcr_dwn[e, s, t] * m.C_FCR_capdown for e in m.EL) \
#                 + sum(m.pfc_fcr_up[f, s, t] * m.C_FCR_capup + m.pfc_fcr_dwn[f, s, t] * m.C_FCR_capdown for f in m.FC) for s in m.SCEN for t in m.TIME)
# m.Obj = pyo.Objective(rule = tot_revenue, sense = pyo.maximize)


#------------------------------------------------------------------------------------------------
# SOLVE
#------------------------------------------------------------------------------------------------

solver = pyo.SolverFactory('gurobi')  # change to "highs" if you don't have Gurobi
# res = solver.solve(m, tee=True)

# solver.options['TimeLimit'] = 600    # time limit for solution [s]
# solver.options['MIPGap'] = 0.05      # acceptable solution gap [p.u. (0.01 = 1%)]

res = solver.solve(m, tee=True, symbolic_solver_labels=True, options={'ResultFile': 'infeasible.ilp'})

if res.solver.termination_condition == pyo.TerminationCondition.infeasible:
    print("Infeasible! Computing IIS...")

    # We must reload the model into gurobipy to ensure names are synced
    m.write("debug_model.lp", io_options={'symbolic_solver_labels': True})
    model_to_check = gp.read("debug_model.lp")

    model_to_check.computeIIS()
    model_to_check.write("final_infeasible_report.ilp")
    print("IIS written to final_infeasible_report.ilp")
```

```python
#------------------------------------------------------------------------------------------------
# EXRACTING RESULTS
#------------------------------------------------------------------------------------------------

# Method 1
# C_ele_vals = np.array([m.C_ele[s, t] for s in m.SCEN for t in m.TIME])
# C_ele_vals = C_ele_vals.reshape(len(m.SCEN), len(m.TIME))

# Method 2
C_ele_vals = np.array([[pyo.value(m.C_ele[s, t]) for t in m.TIME] for s in m.SCEN])

C_ele_sell_vals = np.array([[pyo.value(m.C_ele_sell[s, t]) for t in m.TIME] for s in m.SCEN])
Ppv_vals = np.array([[m.Ppv[s, t] for t in m.TIME] for s in m.SCEN])

# Rigorously, I should extract the values by iterating over the three sets (m.EL, m.SCEN, m.TIME), as shown in the following line
# of code, which results in a 3D array with dimensions (number of ELs, number of scenarios, number of time steps). However, since
# I plan on considering only one EL, for simplicity, I'll build 2D arrays => indexing will be [scenario, timestep]
# pel_vals = np.array([[[m.pel[e, s, t]() for t in m.TIME] for s in m.SCEN] for e in m.EL])
pel_vals = np.array([[m.pel[e, s, t]() for t in m.TIME] for s in m.SCEN for e in m.EL])
hel_vals = np.array([[m.hel[e, s, t]() for t in m.TIME] for s in m.SCEN for e in m.EL])

pel_on_vals = np.array([[m.pel_on[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_on_vals = np.array([[m.xel_on[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_off_vals = np.array([[m.xel_off[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_sb_vals = np.array([[m.xel_sb[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_su_vals = np.array([[m.xel_su[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
pel_fcr_up_vals = np.array([[m.pel_fcr_up[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
pel_fcr_dwn_vals = np.array([[m.pel_fcr_dwn[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])

pb_dch_vals = np.array([[m.pb_dch[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
pb_ch_vals = np.array([[m.pb_ch[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
pb_vals = pb_dch_vals - pb_ch_vals
soc_vals = np.array([[m.soc[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
soc_vals = np.append(np.ones((len(m.SCEN), 1)) * SoC_0['BESS'], soc_vals, axis=1)   # adding the initial SoC value
xb_vals = np.array([[m.xb[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])

hhss_vals = np.array([[m.hhss[hs, s, t].value for t in m.TIME] for s in m.SCEN for hs in m.HSS])
hhss_vals = np.append(np.ones((len(m.SCEN), 1)) * H_hss_0['HSS'], hhss_vals, axis=1)   # adding the initial hydrogen level value

hdem_vals = np.array([[m.hdem[s, t].value for t in m.TIME] for s in m.SCEN])

pfc_vals = np.array([[m.pfc[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])
hfc_vals = np.array([[m.hfc[f, s, t]() for t in m.TIME] for s in m.SCEN for f in m.FC])
pfc_fcr_up_vals = np.array([[m.pfc_fcr_up[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])
pfc_fcr_dwn_vals = np.array([[m.pfc_fcr_dwn[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])

pgrid_p_vals = np.array([[m.pgrid_p[s, t].value for t in m.TIME] for s in m.SCEN])
pgrid_n_vals = np.array([[m.pgrid_n[s, t].value for t in m.TIME] for s in m.SCEN])


#------------------------------------------------------------------------------------------------
# PROCESSING RESULTS
#------------------------------------------------------------------------------------------------

rev_tot = pyo.value(m.Obj)    # objective function value

rev_tot_scen = np.zeros(len(m.SCEN))
rev_h2sell_scen = np.zeros(num_scen)
rev_elecsell_scen = np.zeros(num_scen)
cost_elecbuy_scen = np.zeros(num_scen)
rev_FCR_scen = np.zeros(num_scen)

for s in range(len(m.SCEN)):
    rev_tot_scen[s] = sum(hdem_vals[s, t] * m.C_h2 \
                + pgrid_n_vals[s, t] * D_t * C_ele_sell_vals[s, t] \
                - pgrid_p_vals[s, t] * D_t * C_ele_vals[s, t] \
                + pel_fcr_up_vals[s, t] * D_t * C_FCR_capup + pel_fcr_dwn_vals[s, t] * D_t * C_FCR_capdown \
                + pfc_fcr_up_vals[s, t] * D_t * C_FCR_capup + pfc_fcr_dwn_vals[s, t] * D_t * C_FCR_capdown for t in range(len(m.TIME)))
    
    rev_h2sell_scen[s] = sum(hdem_vals[s, t] * m.C_h2 for t in range(time_window))
    rev_elecsell_scen[s] = sum(pgrid_n_vals[s, t] * D_t * C_ele_sell_vals[s, t] for t in range(time_window))
    cost_elecbuy_scen[s] = sum(pgrid_p_vals[s, t] * D_t * C_ele_vals[s, t] for t in range(time_window))
    rev_FCR_scen[s] = sum((pel_fcr_up_vals[s, t] * C_FCR_capup + pel_fcr_dwn_vals[s, t] * C_FCR_capdown \
                          + pfc_fcr_up_vals[s, t] * C_FCR_capup + pfc_fcr_dwn_vals[s, t] * C_FCR_capdown) * D_t for t in range(time_window))


print(f'Value of the stochastic optimization objective function = {rev_tot:.2f} €')
print(f'Mean value of the total revenue across all scenarios (equal weights) = {rev_tot_scen.mean():.2f} €\n')

print(f'Revenue splits for single scenarios:\n')
for s in range(num_scen):
    print(f'Scenario {s}:')
    print(f'Total revenue = {rev_tot_scen[s]:.2f} €')
    print(f'Revenue from selling hydrogen to the HRS = {rev_h2sell_scen[s]:.2f} €')
    print(f'Revenue from selling excess electricity to the main grid = {rev_elecsell_scen[s]:.2f} €')
    print(f'Cost of buying electricity from the main grid = {cost_elecbuy_scen[s]:.2f} €')
    print(f'Revenue from participation to the FCR service = {rev_FCR_scen[s]:.2f} €\n')
```
