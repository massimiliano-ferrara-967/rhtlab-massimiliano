```python
#------------------------------------------------------------------------------------------------
# PROBLEM DEFINITION IN PYOMO (3/4)
#------------------------------------------------------------------------------------------------

#------------------------------------------------------------------------------------------------
# CONSTRAINTS
#------------------------------------------------------------------------------------------------

# BESS MODEL CONSTRAINTS ------------------------------------------------------------------------
# BESS discharging power
def bess_dch_limits1(m, b, s, t):
    return m.pb_dch[b, s, t] <= m.Cb_mw[b]
m.PbessDischarge1 = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_dch_limits1)
def bess_dch_limits2(m, b, s, t):
    return m.pb_dch[b, s, t] <= m.Cb_mw_max[b] * m.xb[b, s, t]
m.PbessDischarge2 = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_dch_limits2)

# BESS charging power
def bess_ch_limits1(m, b, s, t):
    return m.pb_ch[b, s, t] <= m.Cb_mw[b]
m.PbessCharge1 = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_ch_limits1)
def bess_ch_limits2(m, b, s, t):
    return m.pb_ch[b, s, t] <= m.Cb_mw_max[b] * (1 - m.xb[b, s, t])
m.PbessCharge2 = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = bess_ch_limits2)

# BESS energy modelling (eb[0] = ...; eb[t] = (1 - sig_b_sd) * eb[t-1] + (eta_bch * pb_ch - pb_dch / eta_bdch) * Dt)
def batt_ener_eval(m, b, s, t):
    if t == m.TIME.first():
        return m.eb[b, s, t] == (1 - m.sig_b_sd[b]) * m.SoC_0[b] * m.Cb_mwh[b] + (m.eta_bch[b] * m.pb_ch[b, s, t] - m.pb_dch[b, s, t] / m.eta_bdch[b]) * m.Dt
    else:
        return m.eb[b, s, t] == (1 - m.sig_b_sd[b]) * m.eb[b, s, t-1] + (m.eta_bch[b] * m.pb_ch[b, s, t] - m.pb_dch[b, s, t] / m.eta_bdch[b]) * m.Dt
m.BatteryEnergyEvaluation = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = batt_ener_eval)

# BESS energy limits
def batt_ener_lb(m, b, s, t):
    return m.eb[b, s, t] >= m.alfa_b_dod[b] * m.Cb_mwh[b]
m.BatteryEnergyLB = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = batt_ener_lb)
def batt_ener_ub(m, b, s, t):
    return m.eb[b, s, t] <= m.Cb_mwh[b]
m.BatteryEnergyUB = pyo.Constraint(m.BESS, m.SCEN, m.TIME, rule = batt_ener_ub)

# End SoC condition on the BESS (eb[end] == eb[0])
def batt_ener_end(m, b, s):
    return m.eb[b, s, m.TIME.last()] == m.SoC_0[b] * m.Cb_mwh[b]
m.BatteryEnergyEnd = pyo.Constraint(m.BESS, m.SCEN, rule = batt_ener_end)


# EL MODEL CONSTRAINTS --------------------------------------------------------------------------
# EL working states selection (xel_on[t] + xel_off[t] + xel_sb[t] = 1)
def el_states(m, e, s, t):
    return m.xel_on[e, s, t] + m.xel_off[e, s, t] + m.xel_sb[e, s, t] == 1
m.ELStates = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_states)

# Identification of EL startup (xel_su[0] = 0; xel_su[t] >= xel_off[t-1] + xel_on[t] + xel_sb[t] - 1)
def el_find_startup(m, e, s, t):
    if t == m.TIME.first():
        return m.xel_su[e, s, 0] == 0 
    else:
        return m.xel_su[e, s, t] >= m.xel_off[e, s, t-1] + m.xel_on[e, s, t] + m.xel_sb[e, s, t] - 1
m.ELFindStartup = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_find_startup)

# Hydrogen-power EL piecewise linear characteristic   # REPLACED BY EXPRESSION

# EL power evaluation # REPLACED BY EXPRESSION

# EL active power during on state
def pel_on_limits1(m, e, s, t):
    return m.pel_on[e, s, t] <= m.Cel_max[e] * m.xel_on[e, s, t]
m.ELpelonLimits1 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_on_limits1)
def pel_on_limits2(m, e, s, t):
    return m.pel_on[e, s, t] <= m.Cel[e]
m.ELpelonLimits2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_on_limits2)
def pel_on_limits3(m, e, s, t):
    return m.pel_on[e, s, t] >= m.alfa_el_min[e] * m.Cel[e] - m.Cel_max[e] * (1 - m.xel_on[e, s, t])
m.ELpelonLimits3 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_on_limits3)

# EL active power during stand-by state
def pel_sb_limits1(m, e, s, t):
    return m.pel_sb[e, s, t] <= m.alfa_el_sb[e] * m.Cel_max[e] * m.xel_sb[e, s, t]
m.ELpelsbLimits1 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_sb_limits1)
def pel_sb_limits2(m, e, s, t):
    return m.pel_sb[e, s, t] <= m.alfa_el_sb[e] * m.Cel[e]
m.ELpelsbLimits2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_sb_limits2)
def pel_sb_limits3(m, e, s, t):
    return m.pel_sb[e, s, t] >= m.alfa_el_sb[e] * m.Cel[e] - m.Cel_max[e] * (1 - m.xel_sb[e, s, t])
m.ELpelsbLimits3 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = pel_sb_limits3)

# EL ramp rate constraints
def el_rampup(m, e, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip
    return m.pel[e, s, t+1] - m.pel[e, s, t] <= m.alfa_el_rhopos[e] * m.Cel[e]
m.ELrampup = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_rampup)
def el_rampdown(m, e, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip 
    return m.alfa_el_rhoneg[e] * m.Cel[e] <= m.pel[e, s, t+1] - m.pel[e, s, t]
m.ELrampdown = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_rampdown)


# HSS MODEL CONSTRAINTS -------------------------------------------------------------------------
# Hydrogen tank level modelling
def h2_tank_level(m, hs, s, t):
    if t == m.TIME.first():
        return m.hhss[hs, s, t] == (1 - m.sig_hss_sd[hs]) * m.SoC_hss_0[hs] * m.Chss[hs] + \
            (1 - m.sig_hss_ch[hs]) * (1 - m.sig_hc_el2hss[hs]) * sum(m.hel[e, s, t] for e in m.EL) \
            - sum(m.hfc[f, s, t] for f in m.FC) / (1 - m.sig_hss_dch[hs]) - m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs]))
    else:
        return m.hhss[hs, s, t] == (1 - m.sig_hss_sd[hs]) * m.hhss[hs, s, t-1] + \
            (1 - m.sig_hss_ch[hs]) * (1 - m.sig_hc_el2hss[hs]) * sum(m.hel[e, s, t] for e in m.EL) \
            - sum(m.hfc[f, s, t] for f in m.FC) / (1 - m.sig_hss_dch[hs]) - m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs]))
m.H2TankLevel = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = h2_tank_level)

# Hydrogen tank level limits
def h2_tank_level_lb(m, hs, s, t):
    return m.hhss[hs, s, t] >= m.alfa_hss_min[hs] * m.Chss[hs]
m.H2TankLevelLB = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = h2_tank_level_lb)
def h2_tank_level_ub(m, hs, s, t):
    return m.hhss[hs, s, t] <= m.Chss[hs]
m.H2TankLevelUB = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = h2_tank_level_ub)

# Hydrogen tank level end condition
def h2_tank_level_end(m, hs, s):
    return m.hhss[hs, s, m.TIME.last()] == m.SoC_hss_0[hs] * m.Chss[hs]
m.H2TankLevelEnd = pyo.Constraint(m.HSS, m.SCEN, rule = h2_tank_level_end)

# In reality, since I'm choosing the parameters alfa_fr_hc_el2hss and FR_hc_hss2dem_max such that the maximum output of the EL can always be sent
# to the HSS and the minimum daily hydrogen demand can be drawn from the HSS in a single hour, the following constraints are basically always 
# verified => they can be removed if this helps the convergence of the model.  
# Maximum compressor flow rate el2hss
def max_comprflow_el2hss(m, hs, s, t):
    return sum(m.hel[e, s, t] for e in m.EL) <= m.alfa_fr_hc_el2hss[hs] * sum(m.Cel[e] for e in m.EL)
m.MaxComprFlowEl2hss = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = max_comprflow_el2hss)

# Maximum compressor flow rate hss2dem
def max_comprflow_hss2dem(m, hs, s, t):
    return m.hdem[s, t] / ((1 - m.sig_hss_dch[hs]) * (1 - m.sig_hc_hss2dem[hs])) <= m.FR_hc_hss2dem_max[hs]
m.MaxComprFlowHss2dem = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = max_comprflow_hss2dem)

# TODO: cancellare il vincolo sotto, alternativa semplificata del primo dei due di sopra
# # Maximum compressor flow rate el2hss
# def max_comprflow_el2hss(m, hs, s, t):
#     return sum(m.hel[e, s, t] for e in m.EL) <= m.FR_hc_el2hss_max[hs]
# m.MaxComprFlowEl2hss = pyo.Constraint(m.HSS, m.SCEN, m.TIME, rule = max_comprflow_el2hss)


# FC MODEL CONSTRAINTS --------------------------------------------------------------------------
# Power-hydrogen linear model   # REPLACED BY EXPRESSION

# FC power boundaries
def pfc_on_limits1(m, f, s, t):
    return m.pfc[f, s, t] <= m.Cfc_max[f] * m.xfc_on[f, s, t]
m.FCpfconLimits1 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = pfc_on_limits1)
def pfc_on_limits2(m, f, s, t):
    return m.pfc[f, s, t] <= m.Cfc[f]
m.FCpfconLimits2 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = pfc_on_limits2)
def pfc_on_limits3(m, f, s, t):
    return m.pfc[f, s, t] >= m.alfa_fc_min[f] * m.Cfc[f] - m.Cfc_max[f] * (1 - m.xfc_on[f, s, t])
m.FCpfconLimits3 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = pfc_on_limits3)

# FC ramp rate constraints
def fc_rampup(m, f, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip
    return m.pfc[f, s, t+1] - m.pfc[f, s, t] <= m.alfa_fc_rhopos[f] * m.Cfc[f]
m.FCrampup = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_rampup)
def fc_rampdown(m, f, s, t):
    if t == m.TIME.last():
        return pyo.Constraint.Skip 
    return m.alfa_fc_rhoneg[f] * m.Cfc[f] <= m.pfc[f, s, t+1] - m.pfc[f, s, t]
m.FCrampdown = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_rampdown)


# FCR MODELLING ---------------------------------------------------------------------------------
# EL FCR participation limits based on power capacity, down-regulation
def el_fcr_caplim_down1(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.Cel[e] - m.pel_on[e, s, t]
m.ELFCRCapacityLimDown1 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_down1)
def el_fcr_caplim_down2(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.Cel_max[e] * m.xel_on[e, s, t]
m.ELFCRCapacityLimDown2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_down2)
# EL FCR participation limits based on power capacity, up-regulation
def el_fcr_caplim_up1(m, e, s, t):
    return m.pel_fcr_up[e, s, t] <= m.pel_on[e, s, t] - m.alfa_el_min[e] * m.Cel[e] + m.Cel_max[e] * (1 - m.xel_on[e, s, t])
m.ELFCRCapacityLimUp1 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_up1)
def el_fcr_caplim_up2(m, e, s, t):
    return m.pel_fcr_up[e, s, t] <= m.Cel_max[e] * m.xel_on[e, s, t]
m.ELFCRCapacityLimUp2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_caplim_up2)

# EL FCR participation limits based on ramp rate
def el_fcr_rampratelim_down(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.alfa_el_rhopos[e] * m.Cel[e] * m.DT_FCR_ramp
m.ELFCRRamprateLimDown = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_rampratelim_down)

def el_fcr_rampratelim_up(m, e, s, t):
    return m.pel_fcr_up[e, s, t] <= -m.alfa_el_rhoneg[e] * m.Cel[e] * m.DT_FCR_ramp
m.ELFCRRamprateLimUp = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_rampratelim_up)

# EL FCR participation limit based on available renewable energy
# If I wanted to generalize this constraint to multiple ELs, the left-hand side would have to be the sum over all the ELs of the available
# capacity for FCR. In other words, the constraint would determine the total available energy for FCR by the ELs, which would then have to
# be divided amongst all the ELs (meaning: sum(m.pel_fcr_dwn[e, t] for i in m.EL) <= left-hand side of the constraint below).
def el_fcr_availab_pow(m, e, s, t):
    return m.pel_fcr_dwn[e, s, t] <= m.Ppv[s, t] + sum(m.pb_dch[b, s, t] - m.pb_ch[b, s, t] for b in m.BESS) - sum(m.pel_on[e, s, t] for e in m.EL) \
                                   - sum(m.phc_el2hss[e, s, t] for e in m.EL) - sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS)
m.ELFCRAvailabPower = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_availab_pow)

# EL FCR participation limit based on available hydrogen storage capacity
def el_fcr_availab_h2(m, e, s, t):
    return m.A[e] * m.pel_fcr_dwn[e, s, t] * m.Dt <= sum(m.Chss[hs] - m.hhss[hs, s, t] for hs in m.HSS)
m.ELFCRAvailabH2 = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_availab_h2)

if sym_fcr:
    # EL FCR symmetric capacity evaluation
    def el_fcr_sym(m, e, s, t):
        return m.pel_fcr_dwn[e, s, t] == m.pel_fcr_up[e, s, t]
    m.ELFCRSymmetric = pyo.Constraint(m.EL, m.SCEN, m.TIME, rule = el_fcr_sym)


# FC FCR participation limits based on power capacity, down-regulation
def fc_fcr_caplim_down1(m, f, s, t):
    return m.pfc_fcr_dwn[f, s, t] <= m.pfc[f, s, t] - m.alfa_fc_min[f] * m.Cfc[f] - m.Cfc_max[f] * (1 - m.xfc_on[f, s, t])
m.FCFCRCapacityLimDown1 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_down1)
def fc_fcr_caplim_down2(m, f, s, t):
    return m.pfc_fcr_dwn[f, s, t] <= m.Cfc_max[f] * m.xfc_on[f, s, t]
m.FCFCRCapacityLimDown2 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_down2)
# FC FCR participation limits based on power capacity, up-regulation
def fc_fcr_caplim_up1(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] <= m.Cfc[f] - m.pfc[f, s, t] 
m.FCFCRCapacityLimUp1 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_up1)
def fc_fcr_caplim_up2(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] <= m.Cfc_max[f] * m.xfc_on[f, s, t]
m.FCFCRCapacityLimUp2 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_caplim_up2)

# FC FCR participation limits based on ramp rate
def fc_fcr_rampratelim_down(m, f, s, t):
    return m.pfc_fcr_dwn[f, s, t] <= -m.alfa_fc_rhoneg[f] * m.Cfc[f] * m.DT_FCR_ramp
m.FCFCRRamprateLimDown = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_rampratelim_down)

def fc_fcr_rampratelim_up(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] <= m.alfa_fc_rhopos[f] * m.Cfc[f] * m.DT_FCR_ramp
m.FCFCRRamprateLimUp = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_rampratelim_up)

# FC FCR participation limit based on hydrogen availability
def fc_fcr_availab_h2(m, f, s, t):
    return m.pfc_fcr_up[f, s, t] / ((1-m.sig_hss_dch['HSS']) * m.eta_fc[f] * m.lhv) <= \
        sum(m.hhss[hs, s, t] - m.alfa_hss_min[hs] * m.Chss[hs] for hs in m.HSS)
m.FCFCRAvailabH2 = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_availab_h2)

if sym_fcr:
    # FC FCR symmetric capacity evaluation
    def fc_fcr_sym(m, f, s, t):
        return m.pfc_fcr_dwn[f, s, t] == m.pfc_fcr_up[f, s, t]
    m.FCFCRSymmetric = pyo.Constraint(m.FC, m.SCEN, m.TIME, rule = fc_fcr_sym)


# SYSTEM CONSTRAINTS ----------------------------------------------------------------------------
# Grid power limits
def grid_power_importlim(m, s, t):
    return m.pgrid_p[s, t] <= m.Pgrid_max_import * m.xgrid[s, t]
m.GridPowerImportLim = pyo.Constraint(m.SCEN, m.TIME, rule=grid_power_importlim)

def grid_power_exportlim(m, s, t):
    return m.pgrid_n[s, t] <= m.Pgrid_max_export * (1 - m.xgrid[s, t])
m.GridPowerExportLim = pyo.Constraint(m.SCEN, m.TIME, rule=grid_power_exportlim)


# Green hydrogen constraint
def green_h2(m, s, t):
    return sum(m.pel[e, s, t] for e in m.EL) + sum(m.pb_ch[b, s, t] for b in m.BESS) \
    + sum(m.phc_el2hss[e, s, t] for e in m.EL) + sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS) <= m.Ppv[s, t] + sum(m.pb_dch[b, s, t] for b in m.BESS)
m.GreenHydrogen = pyo.Constraint(m.SCEN, m.TIME, rule = green_h2)


# General power balance
def power_balance(m, s, t):
    return m.pgrid_p[s, t] - m.pgrid_n[s, t] + m.Ppv[s, t] - m.Pd_load[s, t] \
    + sum(m.pb_dch[b, s, t] - m.pb_ch[b, s, t] for b in m.BESS) \
    - sum(m.pel[e, s, t] for e in m.EL) + sum(m.pfc[f, s, t] for f in m.FC) \
    - sum(m.phc_el2hss[e, s, t] for e in m.EL) - sum(m.phc_hss2dem[hs, s, t] for hs in m.HSS) == 0
m.PowerBalance = pyo.Constraint(m.SCEN, m.TIME, rule = power_balance)


# HYDROGEN DEMAND CONSTRAINTS -------------------------------------------------------------------
# Daily limits on hydrogen supply
def h2_daily_supply_bound_low(m, s, d):
    return m.Hsup_min_day <= sum(m.hdem[s, d*24 + h] for h in range(24))
m.H2DailySupplyBoundLow = pyo.Constraint(m.SCEN, m.DAYS, rule = h2_daily_supply_bound_low)
def h2_daily_supply_bound_high(m, s, d):
    return sum(m.hdem[s, d*24 + h] for h in range(24)) <= m.Hsup_max_day
m.H2DailySupplyBoundHigh = pyo.Constraint(m.SCEN, m.DAYS, rule = h2_daily_supply_bound_high)

# Weekly / Monthly hydrogen supply target
def h2_target_supply(m, s, w):
    return sum(m.hdem[s, w*24*7 + h] for h in range(24*7)) == m.Hsup_tgt_week
m.H2TargetSupply = pyo.Constraint(m.SCEN, m.WEEKS, rule = h2_target_supply)
```

```python
#------------------------------------------------------------------------------------------------
# CONSTRAINT ACTIVATION / DEACTIVATION
#------------------------------------------------------------------------------------------------

# Deactivation of the FC-related constraints when fc_active = False -----------------------------
if not fc_active:
    m.FCpfconLimits1.deactivate()
    m.FCpfconLimits2.deactivate()
    m.FCpfconLimits3.deactivate()
    m.FCrampup.deactivate()
    m.FCrampdown.deactivate()

# Deactivation of the FC-FCR-related constraints when fc_active = False or fcr_active = False ---
if not fc_active or not fcr_active:
    m.FCFCRCapacityLimDown1.deactivate()
    m.FCFCRCapacityLimDown2.deactivate()
    m.FCFCRCapacityLimUp1.deactivate()
    m.FCFCRCapacityLimUp2.deactivate()
    m.FCFCRRamprateLimDown.deactivate()
    m.FCFCRRamprateLimUp.deactivate()
    m.FCFCRAvailabH2.deactivate()
    if sym_fcr:
        m.FCFCRSymmetric.deactivate()

# Deactivation of the EL-FCR-related constraints when fcr_active = False ------------------------
if not fcr_active:
    m.ELFCRCapacityLimDown1.deactivate()
    m.ELFCRCapacityLimDown2.deactivate()
    m.ELFCRCapacityLimUp1.deactivate()
    m.ELFCRCapacityLimUp2.deactivate()
    m.ELFCRRamprateLimDown.deactivate()
    m.ELFCRRamprateLimUp.deactivate()
    m.ELFCRAvailabPower.deactivate()
    m.ELFCRAvailabH2.deactivate()
    if sym_fcr:
        m.ELFCRSymmetric.deactivate()

# Deactivation of the BESS-related constraints when bess_active = False -------------------------
if not bess_active:
    m.PbessDischarge1.deactivate()
    m.PbessDischarge2.deactivate()
    m.PbessCharge1.deactivate()
    m.PbessCharge2.deactivate()
    m.BatteryEnergyEvaluation.deactivate()
    m.BatteryEnergyLB.deactivate()
    m.BatteryEnergyUB.deactivate()
    m.BatteryEnergyEnd.deactivate()
```

```python
#------------------------------------------------------------------------------------------------
# PROBLEM DEFINITION IN PYOMO (4/4)
#------------------------------------------------------------------------------------------------

#------------------------------------------------------------------------------------------------
# OBJECTIVE FUNCTION
#------------------------------------------------------------------------------------------------

def tot_revenue(m):
    return -(m.Capex_pv * m.Ppv_rtd * m.CRF + m.Opex_pv * m.Ppv_rtd \
               + (m.Capex_b_mwh * m.Cb_mwh['BESS'] + m.Capex_b_mw * m.Cb_mw['BESS']) * (m.CRF + m.Opex_b_percent) \
               + (m.Capex_el_dir * m.Cel['EL']) * (1 + m.Capex_el_indir_percent) * m.CRF + (m.Capex_el_dir * m.Cel['EL']) * m.Opex_el_percent \
                + (m.Capex_hss * m.Chss['HSS']) * (m.CRF + m.Opex_hss_percent) \
                + (m.Capex_fc * m.Cfc['FC'] * m.CRF + m.Opex_fc * m.Cfc['FC']) \
                + (m.Capex_hc_pu_flux * m.alfa_fr_hc_el2hss['HSS'] * m.Cel['EL'] + m.Capex_hc_pu_flux * m.FR_hc_hss2dem_max['HSS']) * m.CRF) \
                + (1 / m.num_scen) * sum(m.hdem[s, t] * m.C_h2 \
                + m.pgrid_n[s, t] * m.Dt * m.C_ele_sell[s, t] \
                + (m.Pd_load[s, t] - m.pgrid_p[s, t]) * m.Dt * m.C_ele[s, t] \
                + (sum(m.pel_fcr_up[e, s, t] * m.C_FCR_capup + m.pel_fcr_dwn[e, s, t] * m.C_FCR_capdown for e in m.EL) \
                + sum(m.pfc_fcr_up[f, s, t] * m.C_FCR_capup + m.pfc_fcr_dwn[f, s, t] * m.C_FCR_capdown for f in m.FC)) * m.Dt for s in m.SCEN for t in m.TIME)
m.Obj = pyo.Objective(rule = tot_revenue, sense = pyo.maximize)               

# def tot_revenue(m):
#     return -(m.Capex_pv * m.Ppv_rtd * m.CRF + m.Opex_pv * m.Ppv_rtd \
#                + (m.Capex_b_mwh * m.Cb_mwh['BESS'] + m.Capex_b_mw * m.Cb_mw['BESS']) * (m.CRF + m.Opex_b_percent) \
#                + (m.Capex_el_dir * m.Cel['EL']) * (1 + m.Capex_el_indir_percent) * m.CRF + (m.Capex_el_dir * m.Cel['EL']) * m.Opex_el_percent \
#                 + (m.Capex_hss * m.Chss['HSS']) * (m.CRF + m.Opex_hss_percent) \
#                 + (m.Capex_fc * m.Cfc['FC'] * m.CRF + m.Opex_fc * m.Cfc['FC']) \
#                 + (m.Capex_hc_pu_flux * m.FR_hc_el2hss_max['HSS'] + m.Capex_hc_pu_flux * m.FR_hc_hss2dem_max['HSS']) * m.CRF) \
#                 + (1 / m.num_scen) * sum(m.hdem[s, t] * m.C_h2 \
#                 + m.pgrid_n[s, t] * m.Dt * m.C_ele_sell[s, t] \
#                 - m.pgrid_p[s, t] * m.Dt * m.C_ele[s, t] \
#                 + (sum(m.pel_fcr_up[e, s, t] * m.C_FCR_capup + m.pel_fcr_dwn[e, s, t] * m.C_FCR_capdown for e in m.EL) \
#                 + sum(m.pfc_fcr_up[f, s, t] * m.C_FCR_capup + m.pfc_fcr_dwn[f, s, t] * m.C_FCR_capdown for f in m.FC)) * m.Dt for s in m.SCEN for t in m.TIME)
# m.Obj = pyo.Objective(rule = tot_revenue, sense = pyo.maximize)               


#------------------------------------------------------------------------------------------------
# SOLVE
#------------------------------------------------------------------------------------------------

solver = pyo.SolverFactory('gurobi')
# res = solver.solve(m, tee=True)

solver.options['TimeLimit'] = 3600    # time limit for solution [s]
# solver.options['MIPGap'] = 0.005      # acceptable solution gap [p.u. (0.01 = 1%)]

res = solver.solve(m, tee=True, symbolic_solver_labels=True, options={'ResultFile': 'infeasible.ilp'})

if res.solver.termination_condition == pyo.TerminationCondition.infeasible:
    print("Infeasible! Computing IIS...")

    # We must reload the model into gurobipy to ensure names are synced
    m.write("debug_model.lp", io_options={'symbolic_solver_labels': True})
    model_to_check = gp.read("debug_model.lp")

    model_to_check.computeIIS()
    model_to_check.write("final_infeasible_report.ilp")
    print("IIS written to final_infeasible_report.ilp")
```

```python
#------------------------------------------------------------------------------------------------
# EXRACTING RESULTS
#------------------------------------------------------------------------------------------------

C_ele_vals = np.array([[pyo.value(m.C_ele[s, t]) for t in m.TIME] for s in m.SCEN])
C_ele_sell_vals = np.array([[pyo.value(m.C_ele_sell[s, t]) for t in m.TIME] for s in m.SCEN])
Ppv_vals = np.array([[m.Ppv[s, t] for t in m.TIME] for s in m.SCEN])
Pload_vals = np.array([[m.Pd_load[s, t] for t in m.TIME] for s in m.SCEN])

# Rigorously, I should extract the values by iterating over the three sets (m.EL, m.SCEN, m.TIME), as shown in the following line
# of code, which results in a 3D array with dimensions (number of ELs, number of scenarios, number of time steps). However, since
# I plan on considering only one EL, for simplicity, I'll build 2D arrays => indexing will be [scenario, timestep]
# pel_vals = np.array([[[m.pel[e, s, t]() for t in m.TIME] for s in m.SCEN] for e in m.EL])
pel_vals = np.array([[m.pel[e, s, t]() for t in m.TIME] for s in m.SCEN for e in m.EL])
hel_vals = np.array([[m.hel[e, s, t]() for t in m.TIME] for s in m.SCEN for e in m.EL])

pel_on_vals = np.array([[m.pel_on[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_on_vals = np.array([[m.xel_on[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_off_vals = np.array([[m.xel_off[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_sb_vals = np.array([[m.xel_sb[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
xel_su_vals = np.array([[m.xel_su[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
pel_fcr_up_vals = np.array([[m.pel_fcr_up[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])
pel_fcr_dwn_vals = np.array([[m.pel_fcr_dwn[e, s, t].value for t in m.TIME] for s in m.SCEN for e in m.EL])

pb_dch_vals = np.array([[m.pb_dch[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
pb_ch_vals = np.array([[m.pb_ch[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
pb_vals = pb_dch_vals - pb_ch_vals
eb_vals = np.array([[m.eb[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])
eb_vals = np.append(np.ones((len(m.SCEN), 1)) * m.SoC_0['BESS'] * m.Cb_mwh['BESS'].value, eb_vals, axis=1)    # adding the initial energy value
soc_vals = eb_vals / m.Cb_mwh['BESS'].value
xb_vals = np.array([[m.xb[b, s, t].value for t in m.TIME] for s in m.SCEN for b in m.BESS])

hhss_vals = np.array([[m.hhss[hs, s, t].value for t in m.TIME] for s in m.SCEN for hs in m.HSS])
hhss_vals = np.append(np.ones((len(m.SCEN), 1)) * m.SoC_hss_0['HSS'] * m.Chss['HSS'].value, hhss_vals, axis=1)    # adding the initial hydrogen level value

hdem_vals = np.array([[m.hdem[s, t].value for t in m.TIME] for s in m.SCEN])

pfc_vals = np.array([[m.pfc[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])
hfc_vals = np.array([[m.hfc[f, s, t]() for t in m.TIME] for s in m.SCEN for f in m.FC])
pfc_fcr_up_vals = np.array([[m.pfc_fcr_up[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])
pfc_fcr_dwn_vals = np.array([[m.pfc_fcr_dwn[f, s, t].value for t in m.TIME] for s in m.SCEN for f in m.FC])

pgrid_p_vals = np.array([[m.pgrid_p[s, t].value for t in m.TIME] for s in m.SCEN])
pgrid_n_vals = np.array([[m.pgrid_n[s, t].value for t in m.TIME] for s in m.SCEN])


#------------------------------------------------------------------------------------------------
# PROCESSING RESULTS
#------------------------------------------------------------------------------------------------

rev_tot = pyo.value(m.Obj)    # objective function value

# Scenario-independent results
ann_install_costs = m.CRF * (m.Capex_pv * m.Ppv_rtd + (m.Capex_b_mwh * m.Cb_mwh['BESS'].value + m.Capex_b_mw * m.Cb_mw['BESS'].value) \
                + (m.Capex_el_dir * m.Cel['EL'].value) * (1 + m.Capex_el_indir_percent) + (m.Capex_hss * m.Chss['HSS'].value) \
                + (m.Capex_fc * m.Cfc['FC'].value) \
                + (m.Capex_hc_pu_flux * m.alfa_fr_hc_el2hss['HSS'] * m.Cel['EL'].value + m.Capex_hc_pu_flux * m.FR_hc_hss2dem_max['HSS']))
ann_om_costs = m.Opex_pv * m.Ppv_rtd + (m.Capex_b_mwh * m.Cb_mwh['BESS'].value + m.Capex_b_mw * m.Cb_mw['BESS'].value) * m.Opex_b_percent \
                + (m.Capex_el_dir * m.Cel['EL'].value) * m.Opex_el_percent + (m.Capex_hss * m.Chss['HSS'].value) * m.Opex_hss_percent \
                + m.Opex_fc * m.Cfc['FC'].value

# Scenario-dependent results
rev_tot_scen = np.zeros(len(m.SCEN))
rev_h2sell_scen = np.zeros(num_scen)
rev_elecsell_scen = np.zeros(num_scen)
rev_loadpower_scen = np.zeros(num_scen)
rev_FCR_scen = np.zeros(num_scen)

for s in range(len(m.SCEN)):
    rev_tot_scen[s] = -ann_install_costs - ann_om_costs + sum(hdem_vals[s, t] * m.C_h2 \
                + pgrid_n_vals[s, t] * D_t * C_ele_sell_vals[s, t] \
                + (m.Pd_load[s, t] - pgrid_p_vals[s, t]) * D_t * C_ele_vals[s, t] \
                + (pel_fcr_up_vals[s, t] * m.C_FCR_capup + pel_fcr_dwn_vals[s, t] * m.C_FCR_capdown \
                + pfc_fcr_up_vals[s, t] * m.C_FCR_capup + pfc_fcr_dwn_vals[s, t] * m.C_FCR_capdown) * D_t for t in range(len(m.TIME)))
    
    rev_h2sell_scen[s] = sum(hdem_vals[s, t] * m.C_h2 for t in range(time_window))
    rev_elecsell_scen[s] = sum(pgrid_n_vals[s, t] * D_t * C_ele_sell_vals[s, t] for t in range(time_window))
    rev_loadpower_scen[s] = sum((m.Pd_load[s, t] - pgrid_p_vals[s, t]) * D_t * C_ele_vals[s, t] for t in range(time_window))
    rev_FCR_scen[s] = sum((pel_fcr_up_vals[s, t] * m.C_FCR_capup + pel_fcr_dwn_vals[s, t] * m.C_FCR_capdown \
                          + pfc_fcr_up_vals[s, t] * m.C_FCR_capup + pfc_fcr_dwn_vals[s, t] * m.C_FCR_capdown) * D_t for t in range(time_window))


print(f'Value of the stochastic optimization objective function = {rev_tot:.2f} €')
print(f'Mean value of the net revenue across all scenarios (equal weights) = {rev_tot_scen.mean():.2f} €\n')

print(f"Optimal BESS size: {m.Cb_mwh['BESS'].value:.2f} MWh, {m.Cb_mw['BESS'].value:.2f} MW")
print(f"Optimal EL size: {m.Cel['EL'].value:.2f} MW")
print(f"Optimal HSS size: {m.Chss['HSS'].value:.2f} kg")
print(f"Optimal FC size: {m.Cfc['FC'].value:.2f} MW\n")

print(f"Annualized installation costs = {ann_install_costs:.2f} €")
print(f"Annual O&M costs = {ann_om_costs:.2f} €\n\n")

print(f'Revenue splits for the single scenarios:\n')
for s in range(num_scen):
    print(f'Scenario {s}:')
    print(f'Net revenue = {rev_tot_scen[s]:.2f} €')
    print(f'Revenue from selling hydrogen to the HRS = {rev_h2sell_scen[s]:.2f} €')
    print(f'Revenue from selling excess electricity to the main grid = {rev_elecsell_scen[s]:.2f} €')
    print(f'Revenue from powering the load = {rev_loadpower_scen[s]:.2f} €')
    print(f'Revenue from participation to the FCR service = {rev_FCR_scen[s]:.2f} €\n')
```
