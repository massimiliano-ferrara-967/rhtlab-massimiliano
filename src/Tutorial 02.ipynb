```python
# Examples of indexing and slicing
a = np.array([["A", "B", "C", "D"], 
              ["E", "F", "G", "H"],
              ["I", "J", "K", "L"]])
print(a)

print(a[1])
print(a[:, 1])
print(a[1, 1])
print(a[[0, 2], 1])
print(a[[0, 2], [1, 3]])
print(a[1, 1:])
print(a[1, -2:])
```
