```python
# Examples of indexing and slicing
a = np.array([["A", "B", "C", "D"], 
              ["E", "F", "G", "H"],
              ["I", "J", "K", "L"]])
print(a)

print(a[1])
print(a[:, 1])
print(a[1, 1])
print(a[[0, 2], 1])
print(a[[0, 2], [1, 3]])
print(a[1, 1:])
print(a[1, -2:])
```

```python
b = np.array([["M", "N", "O", "P"]])
np.concatenate((a, b), axis=0)
```

```python
# Modifying array elements
a[2, 2] = "X"
print(a)
a[1, :] = ["W", "X", "Y", "Z"]
print(a)
```

```python
a = np.array([14, 23, 32, 41])
b = np.array([5, 4, 3, 2])
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
```

First and second rule of broadcasting

```python
h = np.arange(5).reshape(1, 1, 5)
h
```

```python
h + [10, 20, 30, 40, 50] # same as: h+ [[[10, 20, 30, 40, 50]]]
```

```python
k = np.arange(6).reshape(2, 3)
k
```

```python
k + [[100], [200]] # same as: k + [[100, 100, 100], [200, 200, 200]]
```

```python
k + [100, 200, 300] # after rule 1: [[100, 200, 300]], and after rule 2: [[100, 200, 300], [100, 200, 300]]
```

```python
k + 1000 # same as: k + [[1000, 1000, 1000], [1000, 1000, 1000]]
```

```python
# Third rule of broadcasting
try:
    k + [33, 44]
except ValueError as e:
    print(e)
```

```python
# Conditional operators apply elementwise
m = np.array([20, -5, 30, 40])
m < [15, 16, 35, 36]
```

```python
# They also use broadcasting
m < 25 # equivalent to m < [25, 25, 25, 25]
```

```python
# Conditional operators can be used for indexing
m[m < 25]
```

Other ndarray general operators

```python
a = np.array([[-2.5, 3.1, 7], [10, 11, 12]])
print(a)
for func in (np.square, np.sqrt, np.exp, np.log, np.sign, np.ceil, np.isnan, np.cos):
    print("\n", func.__name__)
    print(func(a))

# The two warnings are due to the fact that sqrt() and log() are undefined for negative numbers, which is why there is a np.nan value in
# the first cell of the output of these two functions.
```

ndarray statistical operators

```python
a = np.array([[-2.5, 3.1, 7], [10, 11, 12]])
for func in (a.min, a.max, a.sum, a.std, a.var):
    print(func.__name__, "=", func())

c = np.arange(24).reshape(2,3,4)
print(c)

print(c.sum(axis=0)) # sum across matrices
print(c.sum(axis=1)) # sum across rows
print(c.sum(axis=2)) # sum across columns
print(c.sum(axis=(0,2))) # sum across matrices and columns
```

```python
a = np.array([1, -2, 3, 4])
b = np.array([2, 8, -1, 7])

print(np.maximum(a, b))
np.copysign(a, b)
```

Interpolation

```python
xp = [1, 2, 3]
fp = [3, 2, 0]

np.interp(2.5, xp, fp)
```

```python
np.interp([0, 1, 1.5, 2.72, 3.14], xp, fp)
```

```python
x = np.linspace(0, 2*np.pi, 10)
y = np.sin(x)
xvals = np.linspace(0, 2*np.pi, 50)
yinterp = np.interp(xvals, x, y)
import matplotlib.pyplot as plt
#%matplotlib notebook
plt.plot(x, y, 'o')
plt.plot(xvals, yinterp, '-x')
plt.show()
```

```python
# Interpolation with periodic x-coordinates
x = [-180, -170, -185, 185, -10, -5, 0, 365]
xp = [190, -190, 350, -350]
fp = [5, 10, 3, 4]
np.interp(x, xp, fp, period=360)
```

```python
# Complex interpolation
x = [1.5, 4.0]
xp = [2, 3, 5]
fp = [1.0j, 0, 2+3j]
np.interp(x, xp, fp)
```
