# Lists

```python
squares = [1, 4, 9, 16, 25, 36, 49, 64]
print(squares)
print(type(squares))

# Items in a list can be of different data types
my_list = [1, 'A', 1.0, True]
print(my_list)

len(squares)
```

```python
print(squares[0]) # indexing returns the item
print(squares[2])

# Negative indexing returns items counting from the end
print(squares[-1])
print(squares[-2])

print(squares[2:5])

print(squares[:3])
print(squares[-3:])

print(squares[::2])      # from start to finish 2 by 2
print(squares[1::2])     # from index 1 to end 2 by 2
print(squares[1:-3:2])   # from index 1 to index -3 (last but three element) 2 by 2

squares[:]    # shallow copy of the list
```

```python
a = [1, 2, 3]
b = a
print(a)
print(b)

a[1] = 4
print(a)
print(b)
```

```python
# Lists are a mutable type, i.e. it is possible to change their content, like concatenation:
squares + [81, 100]
```

```python
squares.append(121) # add 11 squared
squares.append(12 ** 2)  # add 12 squared
squares
```

```python
# Elements in a list can be changed
letters = ['a', 'b', 'c', 'x', 'e', 'f', 'g']
print(letters)

letters[3] = 'd' 
print(letters)

letters[2:5] = ['X']    # replacing a slice with another list (in this case a single element)
print(letters)

letters[2:5] = []
print(letters)

letters[:] = []    # clearing the entire list
print(letters)
```

```python
# Nested lists
a = ['a', 'b', 'c']
n = [1, 2, 3]
x = [a, n]
print(x)

print(x[0])
print(x[0][1])

```

```python
# If example 1
x = 42
if x < 0:
    x = 0
    print('Negative changed to zero')
elif x == 0:
    print('Zero')
elif x == 1:
    print('Single')
else:
    print('More')

# If example 2
x = 42
if (x >= 0) & (x <= 1):
    x = 0
    print('x between 0 and 1')
elif (x >= 3) | (x <= -4):
    print('x gater than 3 or smaller than -4')
else:
    print('other')
```

```python
# While example
# The sum of two elements defines the next
a, b = 0, 1    # example of multiple assignment
while a < 10:
    print(a)
    a, b = b, a+b
```

## Example of shallow vs deep copy

```python
# Assignements do not create copies, but rather references to a given object
A = [1, 2, 3]
B = A
B.append(4)

print(A)
print(B)
```

```python
import copy

A = [1, 2, 3]
# B = A.copy()
B = copy.copy(A)  # The same as `B = A.copy()`
B.append(4)

print(A)
print(B)
```

```python
# Example of shallow copy behavior
A = [[1, 2], ['a', 'b']]
B = copy.copy(A)  # The same as `B = A.copy()`
B.append(4)    # an append applied on the first level affects only B 
print(A)
print(B)

B[0].append(3)    # an append applied on the second level affects both A and B
print(A)
print(B)
```

```python
# Example of deep copy behavior
A = [[1, 2], ['a', 'b']]
B = copy.deepcopy(A)

B.append(4)
print(A)
# [[1, 2], ['a', 'b']]
print(B)
# [[1, 2], ['a', 'b'], 4]

B[0].append(3)
print(A)
# [[1, 2], ['a', 'b']]
print(B)
# [[1, 2, 3], ['a', 'b'], 4]
```
